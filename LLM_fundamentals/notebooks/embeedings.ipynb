```python
import gensim.downloader as api

# Load pre-trained Word2Vec model
model = api.load("word2vec-google-news-300")
```

```python
model['cat']
```

```python
from sklearn.decomposition import PCA
from sklearn.metrics.pairwise import cosine_similarity
import matplotlib.pyplot as plt
# from mpl_toolkits.mplot3d import Axes3D
# import numpy as np

v_cat = model['cat']
v_dog = model['dog']
v_cow = model['cow']
v_parrot = model['parrot']
v_result = v_cat - v_cow + v_dog - .1*v_parrot


words_to_plot = ['cat', 'dog', 'cow', 'parrot', 'result']
vectors_to_plot = [v_cat, v_dog, v_cow, v_parrot, v_result]

pca = PCA(n_components=3)
reduced = pca.fit_transform(vectors_to_plot)
coords = {word: reduced[i] for i, word in enumerate(words_to_plot)}

def cos(a, b):
    return cosine_similarity([a], [b])[0, 0]

fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

for word, vec in coords.items():
    ax.scatter(vec[0], vec[1], vec[2])
    ax.text(vec[0], vec[1], vec[2], word)

result = coords['result']
cat = coords['cat']
dog = coords['dog']

# Result ↔ cat
ax.plot([result[0], cat[0]], [result[1], cat[1]], [result[2], cat[2]], color='red')
mid_rk = (result + cat) / 2.0
ax.text(mid_rk[0], mid_rk[1], mid_rk[2],
        f"sim={cos(v_result, v_cat):.2f}", color='red')

# cat ↔ dog
ax.plot([cat[0], dog[0]], [cat[1], dog[1]], [cat[2], dog[2]], color='red')
mid_kq = (cat + dog) / 2.0
ax.text(mid_kq[0], mid_kq[1], mid_kq[2],
        f"sim={cos(v_cat, v_dog):.2f}", color='red')

# Result ↔ dog
ax.plot([result[0], dog[0]], [result[1], dog[1]], [result[2], dog[2]], color='blue')
mid_rq = (result + dog) / 2.0
ax.text(mid_rq[0], mid_rq[1], mid_rq[2],
        f"sim={cos(v_result, v_dog):.2f}", color='blue')

ax.set_title("3D PCA — Cat - Cow + dog - = Result")
ax.set_xlabel("PCA 1") 
ax.set_ylabel("PCA 2") 
ax.set_zlabel("PCA 3")
plt.show()
```

```python
similar_words = model.most_similar(positive=[v_dog], topn=10)

print("Top 10 palabras mas similares a la operación 'dog':")
for word, similarity in similar_words:
    print(f"{word}: {similarity:.4f}")
```
